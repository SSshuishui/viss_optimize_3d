```python
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
import healpy as hp
```

```python
NSIDE = 4096
print(
    "Approximate resolution at NSIDE {} is {:.2} deg".format(
        NSIDE, hp.nside2resol(NSIDE, arcmin=True) / 60
    )
)
```

```python
NPIX = hp.nside2npix(NSIDE)
print(NPIX)
```

```python
ori = np.loadtxt("/data/zhaox/earth_10Mhz/B_10M_original.txt")
```

```python
hp.mollview(ori, nest=True, title="Mollview image B with NSIDE=4096")
hp.graticule()  # 添加经纬线
plt.show()
```

```python
total_data = np.loadtxt("./3dnoblockage10M/C1day1M.txt")


hp.mollview(total_data, nest=True, title="Mollview image healpix No Blockage with NSIDE 4096")
hp.graticule()  # 添加经纬线
plt.show()
```

```python
# total_data = np.loadtxt("./3dnoblockage10M/C1day1M.txt")

# 读取并累加450个文件
for i in range(2, 451):
    file_path = f"./3dnoblockage10M/C{i}day1M.txt"
    data = np.loadtxt(file_path)
    print(f"{i} loaded")
    
    # 累加数据
    total_data += data

hp.mollview(total_data, nest=True, title="Mollview image healpix No Blockage with NSIDE=4096")
hp.graticule()  # 添加经纬线
plt.show()
```
