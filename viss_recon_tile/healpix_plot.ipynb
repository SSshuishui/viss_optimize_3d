```python
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
import healpy as hp
```

# 1Mhz

```python
NSIDE = 512
print(
    "Approximate resolution at NSIDE {} is {:.2} deg".format(
        NSIDE, hp.nside2resol(NSIDE, arcmin=True) / 60
    )
)
```

```python
NPIX = hp.nside2npix(NSIDE)
print(NPIX)
```

```python
recon = np.fromfile("./out1M/C1day1M.bin", dtype=np.float32)

hp.mollview(recon, 
            title="Mollview Sky Image With NSide=512")
plt.show()
```

# 10Mhz

```python
NSIDE = 4096
print(
    "Approximate resolution at NSIDE {} is {:.2} deg".format(
        NSIDE, hp.nside2resol(NSIDE, arcmin=True) / 60
    )
)
NPIX = hp.nside2npix(NSIDE)
print(NPIX)
```

```python
recon10m = np.fromfile("./out10M/C1day10M.bin", dtype=np.float32)

hp.mollview(recon10m, 
            nest=True, 
            title="Mollview Sky Image With NSide=4096")
plt.show()
```
