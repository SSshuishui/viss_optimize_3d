```python
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
import healpy as hp
```

```python
NSIDE = 4096
print(
    "Approximate resolution at NSIDE {} is {:.2} deg".format(
        NSIDE, hp.nside2resol(NSIDE, arcmin=True) / 60
    )
)
```

```python
NPIX = hp.nside2npix(NSIDE)
print(NPIX)
```

```python
ori = np.loadtxt("/data/zhaox/earth_10Mhz/B_10M.txt")
```

```python
recon = np.loadtxt("./3dblockage10M/C1day10M.txt")
```

```python
print(len(ori))
print(len(recon))
```

```python
fig, axs = plt.subplots(1, 2, figsize=(18, 6))

hp.mollview(
    ori,
    nest=True,
    title="Mollview Original Image",
    sub=(1, 2, 1),
    cmap='viridis',
)
axs[0].axis('off')

hp.mollview(
    recon,
    nest=True,
    title="Mollview Block Recon Image",
    sub=(1, 2, 2),
    cmap='viridis',
)
axs[1].axis('off')

plt.show()
```

```python
norecon = np.loadtxt("./3dnoblockage10M/C1day10M.txt")
```

```python
fig, axs = plt.subplots(1, 2, figsize=(18, 6))

hp.mollview(
    ori,
    nest=True,
    title="Mollview Original Image",
    sub=(1, 2, 1),
    cmap='viridis',
)
axs[0].axis('off')

hp.mollview(
    norecon,
    nest=True,
    title="Mollview No Blockage Image",
    sub=(1, 2, 2),
    cmap='viridis',
)
axs[1].axis('off')

plt.show()
```
